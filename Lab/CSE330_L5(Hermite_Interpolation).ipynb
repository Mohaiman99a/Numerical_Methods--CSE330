```python
import numpy as np
from numpy.polynomial import Polynomial
```

```python
def hermit(x, y, y_prime):
    f = 0.0
    for i in range(len(x)):
        f += (y[i] * h(i, x)) + (y_prime[i] * h_hat(i, x))

    return f
```

```python
def h(k, x):
    l_k = l(k, x)
    l_k_sqr = l_k ** 2
    l_k_prime = l_k.deriv(1)


    coeff = [1 + 2 * x[k] * l_k_prime(x[k]), -2 * l_k_prime(x[k])]
    p = Polynomial(coeff)

    return p * l_k_sqr
```

```python
def h_hat(k, x):
    l_k = l(k, x)
    l_k_sqr = l_k ** 2

    coeff = [-x[k], 1]
    p = Polynomial(coeff)

    return p * l_k_sqr
```

```python
def l(k, x):
    l_k = Polynomial([1])
    for i in range(len(x)):
        if i != k:
            term = Polynomial([-x[i], 1]) / (x[k] - x[i])
            l_k = l_k * term
    return l_k
```

```python
p = Polynomial([1, 2, 3])
x = np.array([2.5, 3, 6])
y = p(x)
p_prime = p.deriv(1)
y_prime = p_prime(x)
f = hermit(x, y, y_prime)
print(f)
```
