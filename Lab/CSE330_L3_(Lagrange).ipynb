```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# (a)
def l(k, x, data_x):
    l_k = 1.0
    for i in range(len(data_x)):
        if i != k:
            l_k *= (x - data_x[i]) / (data_x[k] - data_x[i])
    return l_k



def lagrange(x_arr, data_x, data_y):
    degree = len(data_x) - 1
    p_x_arr = np.zeros(len(x_arr))

    for i in range(len(x_arr)):
        for j in range(degree + 1):
            p_x_arr[i] += data_y[j] * l(j, x_arr[i], data_x)

    return p_x_arr
```

```python
# (b)
data_x = [1, 2, 3]
data_y = np.sin(data_x)

test_x = [1.5, 2.5]
result = lagrange(test_x, data_x, data_y)
print(f"Evaluated values: {result}")


plt.plot(test_x, result)
plt.show()
```
