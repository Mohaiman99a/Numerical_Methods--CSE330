```python
import numpy as np
from numpy.polynomial import Polynomial
```

```python
def forward_diff(f, h, x):
    return (f(x+h) - f(x)) / h
```

```python
def backward_diff(f, h, x):
    return (f(x)-f(x-h)) /h
```

```python
def central_diff(f, h, x):
    return (f(x+h)-f(x-h)) / (2*h)
```

# **Richardson Extrapolation**

```python
def dh(f, h, x):
    return (f(x+h)-f(x-h)) / (2*h)
```

```python
def dh1(f, h, x):
    return ((4*dh(f,h/2,x))-dh(f,h,x))/3
```

```python
def error_2(f, hs, x_i):
    f_prime = f.deriv(1)
    Y_actual = f_prime(x_i)

    diff_error = []
    diff2_error = []

    for h in hs:
        diff_error.append(dh(f,h,x_i)-Y_actual)
        diff2_error.append(dh1(f,h,x_i)-Y_actual)

    return diff_error, diff2_error
```
