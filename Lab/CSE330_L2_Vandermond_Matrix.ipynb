```python
import numpy as np
from numpy.polynomial import Polynomial
```

```python
def get_poly(data_x, data_y):
    n_nodes = len(data_x)
    V = np.zeros((n_nodes, n_nodes))

    for i in range(n_nodes):
        for j in range(n_nodes):
            V[i][j] = data_x[i] ** j

    V_inv = np.linalg.pinv(V)
    a = np.dot(V_inv, data_y)
    p = Polynomial(a)

    return p
```

```python
# a
data_x = [1, 2, 3]
data_y = [1, 4, 9]

p = get_poly(data_x, data_y)
```

```python
degree= p.degree()
coef= p.coef

print(f"Degree: {degree}")
print(f"Coefficients: {p.coef}")
```

```python
abs_avg = np.mean(np.abs(p.coef))
print(abs_avg)
```

```python
x_list = [1.5, 2.5, 3.5]
y_list = p(x_list)

print(y_list)
```
