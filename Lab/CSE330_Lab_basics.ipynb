```python
import numpy as np
from numpy.polynomial import Polynomial
import matplotlib.pyplot as plt
```

```python
#a

f = Polynomial([4, -8, 2, -5, 3])    #Lowest to highest degree
val_a = 2 * f(2) - 5
print(f"a) 2 * f(2) - 5 = {val_a}")
```

```python
#b

f_prime = f.deriv(1)
f_double_prime = f.deriv(2)

val_b = f_double_prime(-1) + 3 * f_prime(-1)
print(f"b) f''(-1) + 3 * f'(-1) = {val_b}")
```

```python
#c

custom_val = 2 * f(3) + 1
print(f"c) 2 * f(3) + 1 = {custom_val}")
```

```python
#d

x = np.linspace(-3, 3, 100)
y = f(x)

plt.plot(x, y)
plt.title('Graph of f(x)')
plt.show()
```

```python
#e

x = np.linspace(-3, 3, 100)
y= f(x)
plt.plot(x, y, color='red', label='f(x)')

y2= f.deriv(2)(x)
plt.plot(x, y2, color='green', label="f''(x)")
plt.show()
```
